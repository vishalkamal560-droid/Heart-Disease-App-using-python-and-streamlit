```python
# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

print("Libraries Imported Successfully")
```

```python
# ==========================================
# LOAD DATASET
# ==========================================

data = pd.read_csv(
    '/content/sample_data/Coranary Artery Disease.csv'
)

print("Dataset Shape:")
print(data.shape)

print(data.head())
```

```python
# ==========================================
# DATASET INFORMATION
# ==========================================

print(data.info())

print("\nColumns:")
print(data.columns)

print("\nMissing Values:")
print(data.isnull().sum())
```

```python
# ==========================================
# RENAME COLUMNS
# ==========================================

data = data.rename(columns={

    'Age':'Age',

    'Sex':'Gender',

    'BMI':'Body_Mass_Index',

    'DM':'Diabetes_Mellitus',

    'HTN':'Hypertension',

    'Current Smoker':'Current_Smoker',

    'EX-Smoker':'Ex_Smoker',

    'FH':'Family_History',

    'Obesity':'Obesity',

    'CRF':'Chronic_Renal_Failure',

    'CVA':'Cerebrovascular_Accident',

    'Cath':'CAD_Target'
})

print(data.columns)
```

```python
# ==========================================
# OBJECT TO NUMERIC
# ==========================================

# Gender
data['Gender'] = data['Gender'].map({

    'M':1,

    'F':0
})

# Y/N Columns
yn_columns = [

    'Current_Smoker',

    'Ex_Smoker',

    'Family_History',

    'Chronic_Renal_Failure',

    'Cerebrovascular_Accident'
]

for col in yn_columns:

    data[col] = data[col].map({

        'Y':1,

        'N':0
    })

# Target
data['CAD_Target'] = data['CAD_Target'].map({

    'Cad':1,

    'Normal':0
})

print(data.head())
```

```python
# ==========================================
# HANDLE MISSING VALUES
# ==========================================

num_cols = data.select_dtypes(
    include=np.number
).columns

for col in num_cols:

    data[col] = data[col].fillna(
        data[col].median()
    )

print(data.isnull().sum())
```

```python
# ==========================================
# REMOVE DUPLICATES
# ==========================================

print("Duplicates Before:")
print(data.duplicated().sum())

data = data.drop_duplicates()

print("Duplicates After:")
print(data.duplicated().sum())
```

```python
# ==========================================
# SAFE FEATURES
# ==========================================

selected_features = [

    'Age',

    'Gender',

    'Body_Mass_Index',

    'Diabetes_Mellitus',

    'Hypertension',

    'Current_Smoker',

    'Ex_Smoker',

    'Family_History',

    'Obesity',

    'Chronic_Renal_Failure',

    'Cerebrovascular_Accident',

    'CAD_Target'
]

cad_data = data[selected_features]

print(cad_data.head())
```

```python
# ==========================================
# FEATURE TARGET SPLIT
# ==========================================

X = cad_data.drop(
    'CAD_Target',
    axis=1
)

y = cad_data['CAD_Target']

print(X.shape)
print(y.shape)
```

```python
# ==========================================
# TRAIN TEST SPLIT
# ==========================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.2,

    stratify=y,

    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# Convert 'Obesity' column to numerical before scaling
X_train['Obesity'] = X_train['Obesity'].map({'Y': 1, 'N': 0})
X_test['Obesity'] = X_test['Obesity'].map({'Y': 1, 'N': 0})

# Fit ONLY on training data

X_train = scaler.fit_transform(X_train)

# Transform testing data

X_test = scaler.transform(X_test)

print("Scaling Completed")
```

```python
from imblearn.over_sampling import SMOTE
import pandas as pd
from sklearn.impute import SimpleImputer # Import SimpleImputer

print("Before SMOTE:")
print(y_train.value_counts())

# Apply SMOTE ONLY on training data

smote = SMOTE(random_state=42)

# Impute NaN values in X_train before applying SMOTE
# The data is already scaled, so median imputation is a reasonable strategy.
imputer = SimpleImputer(strategy='median')
X_train_imputed = imputer.fit_transform(X_train)

X_train, y_train = smote.fit_resample(
    X_train_imputed, # Use the imputed data
    y_train
)

print("After SMOTE:")
print(pd.Series(y_train).value_counts())
```

```python
# ==========================================
# CROSS VALIDATION
# ==========================================

from sklearn.model_selection import cross_val_score

from sklearn.ensemble import RandomForestClassifier

rf_cv = RandomForestClassifier(

    n_estimators=100,

    max_depth=5,

    min_samples_split=10,

    min_samples_leaf=5,

    random_state=42
)

cv_scores = cross_val_score(

    rf_cv,

    X_train,

    y_train,

    cv=5,

    scoring='accuracy'
)

print("Cross Validation Scores:")
print(cv_scores)

print("\nAverage Accuracy:")
print(cv_scores.mean())
```

```python
# ==========================================
# TARGET DISTRIBUTION
# ==========================================

plt.figure(figsize=(6,4))

sns.countplot(

    x='CAD_Target',

    data=cad_data
)

plt.title('CAD Target Distribution')

plt.show()
```

```python
# ==========================================
# CORRELATION HEATMAP
# ==========================================

# Convert 'Obesity' column to numerical in cad_data for correlation
cad_data['Obesity'] = cad_data['Obesity'].map({'Y': 1, 'N': 0})

# Impute any remaining NaN values in cad_data to ensure all data is numeric for correlation
# This addresses NaNs in columns like 'Gender', 'Current_Smoker', etc., which are also in cad_data
for col in cad_data.select_dtypes(include=['float64']).columns:
    if cad_data[col].isnull().any():
        cad_data[col] = cad_data[col].fillna(cad_data[col].median())

plt.figure(figsize=(12,8))

sns.heatmap(
    cad_data.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
# ==========================================
# SUPPORT VECTOR MACHINE
# ==========================================

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
# imputer is already defined and fitted in a previous cell (Kv_vo3952V-8)

svm_model = SVC(

    probability=True,

    C=1,

    kernel='rbf',

    random_state=42
)

# Train Model

svm_model.fit(

    X_train, # Use the globally updated X_train which is already imputed and resampled by SMOTE

    y_train
)

# Prediction

# Impute X_test using the same imputer fitted on X_train
X_test_imputed = imputer.transform(X_test)
svm_pred = svm_model.predict(X_test_imputed)

# Accuracy

svm_acc = accuracy_score(

    y_test,

    svm_pred
)

print("SVM Accuracy:")
print(svm_acc)
```

```python
# ==========================================
# KNN MODEL
# ==========================================

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score # Ensure accuracy_score is imported

knn_model = KNeighborsClassifier(

    n_neighbors=7
)

# Train

knn_model.fit(

    X_train, # X_train is already imputed and resampled by SMOTE

    y_train
)

# Prediction

# Use the already imputed X_test_imputed for prediction
knn_pred = knn_model.predict(X_test_imputed)

# Accuracy

knn_acc = accuracy_score(

    y_test,

    knn_pred
)

print("KNN Accuracy:")
print(knn_acc)
```

```python
# ==========================================
# DECISION TREE
# ==========================================

from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(

    max_depth=3,

    min_samples_split=20,

    min_samples_leaf=10,

    random_state=42
)

# Train

dt_model.fit(

    X_train,

    y_train
)

# Prediction

# Use the already imputed X_test_imputed for prediction
dt_pred = dt_model.predict(X_test_imputed)

# Accuracy

dt_acc = accuracy_score(

    y_test,

    dt_pred
)

print("Decision Tree Accuracy:")
print(dt_acc)
```

```python
# ==========================================
# RANDOM FOREST
# ==========================================

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(

    n_estimators=200,

    max_depth=5,

    min_samples_split=10,

    min_samples_leaf=5,

    random_state=42
)

# Train

rf_model.fit(

    X_train,

    y_train
)

# Prediction

# Use the already imputed X_test_imputed for prediction
rf_pred = rf_model.predict(X_test_imputed)

# Accuracy

rf_acc = accuracy_score(

    y_test,

    rf_pred
)

print("Random Forest Accuracy:")
print(rf_acc)
```

```python
# ==========================================
# XGBOOST
# ==========================================

from xgboost import XGBClassifier

xgb_model = XGBClassifier(

    n_estimators=100,

    max_depth=3,

    learning_rate=0.03,

    subsample=0.8,

    colsample_bytree=0.8,

    random_state=42
)

# Train

xgb_model.fit(

    X_train,

    y_train
)

# Prediction

# Use the already imputed X_test_imputed for prediction
xgb_pred = xgb_model.predict(X_test_imputed)

# Accuracy

xgb_acc = accuracy_score(

    y_test,

    xgb_pred
)

print("XGBoost Accuracy:")
print(xgb_acc)
```

```python
# ==========================================
# ADABOOST
# ==========================================

from sklearn.ensemble import AdaBoostClassifier

ada_model = AdaBoostClassifier(

    n_estimators=100,

    learning_rate=0.05,

    random_state=42
)

# Train

ada_model.fit(

    X_train,

    y_train
)

# Prediction

# Use the already imputed X_test_imputed for prediction
ada_pred = ada_model.predict(X_test_imputed)

# Accuracy

ada_acc = accuracy_score(

    y_test,

    ada_pred
)

print("AdaBoost Accuracy:")
print(ada_acc)
```

```python
# ==========================================
# GRID SEARCH CV
# ==========================================

from sklearn.model_selection import GridSearchCV

param_grid = {

    'n_estimators': [100, 200],

    'max_depth': [3, 5, 7],

    'min_samples_split': [5, 10],

    'min_samples_leaf': [2, 5]
}

grid_search = GridSearchCV(

    RandomForestClassifier(random_state=42),

    param_grid,

    cv=5,

    scoring='accuracy',

    n_jobs=-1
)

# Train

grid_search.fit(

    X_train,

    y_train
)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Accuracy:")
print(grid_search.best_score_)
```

```python
# ==========================================
# ANN MODEL
# ==========================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Create Model

ann_model = Sequential()

# Input Layer

ann_model.add(

    Dense(

        64,

        activation='relu',

        input_dim=X_train.shape[1]
    )
)

ann_model.add(
    Dropout(0.3)
)

# Hidden Layer

ann_model.add(

    Dense(

        32,

        activation='relu'
    )
)

ann_model.add(
    Dropout(0.3)
)

# Output Layer

ann_model.add(

    Dense(

        1,

        activation='sigmoid'
    )
)

# Compile Model

ann_model.compile(

    optimizer='adam',

    loss='binary_crossentropy',

    metrics=['accuracy']
)

# Early Stopping

early_stop = EarlyStopping(

    monitor='val_loss',

    patience=5,

    restore_best_weights=True
)

# Train Model

history = ann_model.fit(

    X_train,

    y_train,

    validation_split=0.2,

    epochs=50,

    batch_size=16,

    callbacks=[early_stop],

    verbose=1
)

print("ANN Training Completed")
```

```python
# ==========================================
# HYBRID MODEL
# ==========================================

# RF Probability Features

rf_train_prob = rf_model.predict_proba(X_train)

rf_test_prob = rf_model.predict_proba(X_test_imputed)

# Create Hybrid ANN

hybrid_model = Sequential()

hybrid_model.add(

    Dense(

        16,

        activation='relu',

        input_dim=rf_train_prob.shape[1]
    )
)

hybrid_model.add(

    Dense(

        8,

        activation='relu'
    )
)

hybrid_model.add(

    Dense(

        1,

        activation='sigmoid'
    )
)

# Compile

hybrid_model.compile(

    optimizer='adam',

    loss='binary_crossentropy',

    metrics=['accuracy']
)

# Train

hybrid_history = hybrid_model.fit(

    rf_train_prob,

    y_train,

    validation_split=0.2,

    epochs=20,

    batch_size=8,

    verbose=1
)

print("Hybrid Model Completed")
```
